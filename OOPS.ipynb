Initialize a class person with the constructor to initialize the name and the adhaar number, children and age should set as 0. We have to define get_name() to print the name field, define get_adhaar() to print adhaar number, define set_age to assign a new value to age field also define senior_citizen() which checks for the age>60. Finally create an object for the class and call the respective methods

```python
class Person:
  def __init__(self,name,aadhar_number,age,children):
    self.name = name
    self.aadhar_number = aadhar_number
    self.age = 0
    self.children = 0
  def get_name(self):
    print(f"name : {self.name}")
  def get_aadhar(self):
    print(f"aadhar number : {self.aadhar_number}")
  def set_age(self):
    self.age = age
  def senior_citizen(self):
    if self.age > 60:

        print(f"{self.name} is a senior citizen")
    else:
        print(f"{self.name} is not a senior citizen")



name= input("enter your name:")
aadhar_number= int(input("enter your aadhar number:"))
age= int(input("enter your age:"))
print("-"*8,"verification details","-"*8)
p1= Person(name,aadhar_number,age,children=0)

p1.get_name()
p1.get_aadhar()
p1.set_age()
p1.senior_citizen()
```

Create a class circle with constructor intialising the radius R,another functio area to calculate the area of the circle , another function circumfernce to calculate to calculate the circumference of the circle

```python
class Circle:
  def __init__(self,radius):
    self.radius = radius
  def area(self):
    return 3.14*self.radius**2
  def circumference(self):
    return 2*3.14*self.radius
radius = int(input("enter the radius of the circle:"))
c = Circle(radius)
print(c.area())
print(c.circumference())

```

```python
class Num1:
  def __init__(self,a,b):
     self.a = a
     self.b = b

  def show_values():
    print("a=",a)
    print("b=",b)
class Num2(Num1):
  def show_sum(self):
    print("sum=",self.a+self.b)
  def show_product(self):
    print("product=",self.a*self.b)
oc = Num2(10,20)
oc.show_sum()
oc.show_product()


```

there are 2 classes student and master student,
 the class student has 2 member variables id and mark.there is a constructor intialise the variables  and a function show results which will check the mark for pass and fail.if the mark is greater than equal to 40, otherwise it is fail.similarly in the class masterstudent contain a function show result that will check whether the mark is grater than equal to 50.if that is  the case then ita pass otherwise it is a fail.intialise the student object and a master student object and call the functions respectively

```python
class Student:
  def __init__(self,id,mark):
    self.id = id
    self.mark = mark
  def show_result(self):
    if self.mark >= 40:
      print("The student passed")
    else:
      print("The student failed")
class MasterStudent(Student):

  def show_result(self):
    super().show_result()
    if self.mark >= 50:
      print("The student passed")
    else:
      print("The student failed")
oc=MasterStudent(1,45)
oc.show_result()
```

```python
#polymorphism

class Polygon:
  def __init__(self):
    pass
  def area(self):
    pass
  def perimeter(self):
    pass
class Rectangle(Polygon):
  def __init__(self,length,breadth):
    self.length = length
    self.breadth = breadth
  def area(self):
    return self.length*self.breadth
  def perimeter(self):
    return self(2*(self.length*self.breadth))
class Triangle(Polygon):
  def __init__(self,p,q,r):
    self.p = p
    self.q = q
    self.r = r
  def area(self):
    s = (self.p+self.q+self.r)/2
    return (s*(s-self.p)*(s-self.q)*(s-self.r))**0.5
  def perimeter(self):
    return self.p+self.q+self.r
r = Rectangle(10,20)
t = Triangle(10,20,30)
print("Area of rectangle:",r.area())
print("Perimeter of rectangle:",r.perimeter())
print("Area of triangle:",t.area())
print("Perimeter of triangle:",t.perimeter())
```
